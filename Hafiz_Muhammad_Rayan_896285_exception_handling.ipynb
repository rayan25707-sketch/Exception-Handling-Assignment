## Create multiple custom exception classes for an online payment system. Handle invalid payment amount, insufficient balance, and unsupported payment method as different business errors.

```python
class InvalidPaymentAmountError(Exception):
  pass


class InsufficientBalanceError(Exception):
  pass


class UnsupportedPaymentMethodError(Exception):
  pass


def process_payment(method, amount, balance):
  supported_methods = {"card", "wallet", "bank_transfer"}
  if method.lower() not in supported_methods:
    raise UnsupportedPaymentMethodError(
        f"Payment method '{method}' is not supported."
    )
  if amount <= 0:
    raise InvalidPaymentAmountError("Payment amount must be greater than zero.")
  if amount > balance:
    raise InsufficientBalanceError("Insufficient balance for this payment.")
  return balance - amount


try:
  rem = process_payment("wallet", 2500, 5000)
  print(f"Payment successful! Remaining balance: Rs. {rem}")
  process_payment("cash_on_drone", 1000, 5000)
except (
    InvalidPaymentAmountError,
    InsufficientBalanceError,
    UnsupportedPaymentMethodError,
) as e:
  print(f"Payment failed: {e}")
```

## Create a class-based bank account system. Handle deposits and withdrawals using custom exceptions for invalid transaction amounts and insufficient balance.

```python
class InvalidTransactionAmountError(Exception):
  pass


class InsufficientBalanceError(Exception):
  pass


class BankAccount:

  def __init__(self, owner, balance=0.0):
    self.owner = owner
    self.balance = float(balance)

  def deposit(self, amount):
    if amount <= 0:
      raise InvalidTransactionAmountError(
          "Deposit amount must be greater than zero."
      )
    self.balance += amount
    return self.balance

  def withdraw(self, amount):
    if amount <= 0:
      raise InvalidTransactionAmountError(
          "Withdrawal amount must be greater than zero."
      )
    if amount > self.balance:
      raise InsufficientBalanceError("Insufficient balance for withdrawal.")
    self.balance -= amount
    return self.balance


acc = BankAccount("Ali", 5000)
try:
  acc.deposit(2000)
  acc.withdraw(3000)
  print(f"Updated balance: Rs. {acc.balance}")
  acc.withdraw(10000)
except (InvalidTransactionAmountError, InsufficientBalanceError) as e:
  print(f"Bank Error: {e}")
```

## Create an e-commerce order system that handles invalid products, invalid quantities, insufficient stock, and invalid payments. Use custom exceptions for business-rule failures.

```python
class InvalidProductError(Exception):
  pass


class InvalidQuantityError(Exception):
  pass


class InsufficientStockError(Exception):
  pass


class InvalidPaymentError(Exception):
  pass


def place_order(product_name, qty, payment_valid, inventory):
  if product_name not in inventory:
    raise InvalidProductError(f"Product '{product_name}' does not exist.")
  if not isinstance(qty, int) or qty <= 0:
    raise InvalidQuantityError("Quantity must be a positive integer.")
  item = inventory[product_name]
  if qty > item["stock"]:
    raise InsufficientStockError(
        f"Only {item['stock']} units of '{product_name}' in stock."
    )
  if not payment_valid:
    raise InvalidPaymentError("Payment authorization failed.")
  item["stock"] -= qty
  return item["price"] * qty


store_inventory = {
    "Headphones": {"price": 3500, "stock": 10},
    "USB Cable": {"price": 600, "stock": 25},
}

try:
  total = place_order("Headphones", 2, True, store_inventory)
  print(f"Order placed! Total amount: Rs. {total}")
  place_order("Headphones", 50, True, store_inventory)
except (
    InvalidProductError,
    InvalidQuantityError,
    InsufficientStockError,
    InvalidPaymentError,
) as e:
  print(f"Order Error: {e}")
```

## ### University Course Registration System

### Goal:
Create a program that allows students to register for university courses.

### Steps:
- Store students and courses.
- Check whether the student exists.
- Check whether the course exists.
- Prevent duplicate registration.
- Check course capacity.
- Check prerequisites.
- Create and handle suitable custom exceptions.

### Example:
Student ID: 102
Course: Python
→ Registration successful

```python
class StudentNotFoundError(Exception):
  pass


class CourseNotFoundError(Exception):
  pass


class DuplicateRegistrationError(Exception):
  pass


class CourseFullError(Exception):
  pass


class PrerequisiteNotMetError(Exception):
  pass


students = {
    101: {"name": "Sara", "completed": ["Intro CS"]},
    102: {"name": "Ali", "completed": ["Intro CS"]},
}
courses = {
    "Python": {
        "capacity": 2,
        "enrolled": [101],
        "prerequisites": ["Intro CS"],
    },
    "AI": {"capacity": 1, "enrolled": [], "prerequisites": ["Python"]},
}


def register_course(student_id, course_name):
  if student_id not in students:
    raise StudentNotFoundError(f"Student ID {student_id} not found.")
  if course_name not in courses:
    raise CourseNotFoundError(f"Course '{course_name}' does not exist.")
  course = courses[course_name]
  student = students[student_id]
  if student_id in course["enrolled"]:
    raise DuplicateRegistrationError(
        f"Student {student_id} is already registered in {course_name}."
    )
  if len(course["enrolled"]) >= course["capacity"]:
    raise CourseFullError(f"Course '{course_name}' has reached full capacity.")
  for prereq in course["prerequisites"]:
    if prereq not in student["completed"]:
      raise PrerequisiteNotMetError(f"Missing prerequisite: {prereq}")
  course["enrolled"].append(student_id)
  return "Registration successful"


try:
  print(f"Student ID: 102\nCourse: Python\n-> {register_course(102, 'Python')}")
  print(register_course(102, "AI"))
except (
    StudentNotFoundError,
    CourseNotFoundError,
    DuplicateRegistrationError,
    CourseFullError,
    PrerequisiteNotMetError,
) as e:
  print(f"Registration failed: {e}")
```

## ### Hospital Appointment System

### Goal:
Create a program that manages hospital appointments.

### Steps:
- Store patients and doctors.
- Validate patient and doctor information.
- Check doctor availability.
- Prevent duplicate appointment times.
- Create custom exceptions for important business errors.
- Handle invalid input without crashing.

### Example:
Patient ID: P102
Doctor: D05
Time: 10:30 AM
→ Appointment booked successfully

```python
class InvalidEntityError(Exception):
  pass


class DoctorUnavailableError(Exception):
  pass


class DuplicateAppointmentError(Exception):
  pass


patients = {"P101": "Ayesha", "P102": "Bilal"}
doctors = {
    "D01": {"name": "Dr. Ahmed", "slots": ["09:00 AM", "10:30 AM"]},
    "D05": {"name": "Dr. Sana", "slots": ["10:30 AM", "02:00 PM"]},
}
booked_appointments = set()


def book_appointment(patient_id, doctor_id, time_slot):
  if patient_id not in patients or doctor_id not in doctors:
    raise InvalidEntityError("Invalid patient ID or doctor ID.")
  doc = doctors[doctor_id]
  if time_slot not in doc["slots"]:
    raise DoctorUnavailableError(
        f"{doc['name']} is not available at {time_slot}."
    )
  appt_key = (doctor_id, time_slot)
  if appt_key in booked_appointments:
    raise DuplicateAppointmentError(
        f"Slot {time_slot} with {doc['name']} is already booked."
    )
  booked_appointments.add(appt_key)
  return "Appointment booked successfully"


try:
  status = book_appointment("P102", "D05", "10:30 AM")
  print(
      f"Patient ID: P102\nDoctor: D05\nTime: 10:30 AM\n-> {status}"
  )
  book_appointment("P101", "D05", "10:30 AM")
except (
    InvalidEntityError,
    DoctorUnavailableError,
    DuplicateAppointmentError,
) as e:
  print(f"Booking error: {e}")
```

## ### Digital Wallet System

### Goal:
Create a class-based digital wallet application.

### Steps:
- Validate wallet ID and PIN.
- Handle inactive accounts.
- Validate transfer amounts.
- Check balance.
- Enforce a transfer limit.
- Create multiple custom exceptions.
- Make sure failed transfers do not change wallet balances.

### Example:
Transfer: Rs. 2,000
→ Transfer successful

```python
class WalletAuthError(Exception):
  pass


class InactiveWalletError(Exception):
  pass


class InvalidTransferAmountError(Exception):
  pass


class InsufficientWalletBalanceError(Exception):
  pass


class TransferLimitExceededError(Exception):
  pass


class DigitalWallet:

  def __init__(self, wallet_id, pin, balance, active=True, daily_limit=10000):
    self.wallet_id = wallet_id
    self.pin = pin
    self.balance = float(balance)
    self.active = active
    self.daily_limit = float(daily_limit)

  def transfer(self, entered_pin, amount):
    if entered_pin != self.pin:
      raise WalletAuthError("Invalid PIN.")
    if not self.active:
      raise InactiveWalletError("Wallet account is inactive.")
    if amount <= 0:
      raise InvalidTransferAmountError("Transfer amount must be positive.")
    if amount > self.daily_limit:
      raise TransferLimitExceededError(
          f"Amount exceeds daily limit of Rs. {self.daily_limit}."
      )
    if amount > self.balance:
      raise InsufficientWalletBalanceError("Insufficient wallet balance.")
    self.balance -= amount
    return self.balance


wallet = DigitalWallet("W100", "1234", 8000, active=True, daily_limit=5000)
try:
  new_bal = wallet.transfer("1234", 2000)
  print(
      f"Transfer: Rs. 2,000\n-> Transfer successful (Balance: Rs."
      f" {new_bal:.2f})"
  )
  wallet.transfer("1234", 7000)
except (
    WalletAuthError,
    InactiveWalletError,
    InvalidTransferAmountError,
    InsufficientWalletBalanceError,
    TransferLimitExceededError,
) as e:
  print(f"Transfer failed (Balance unchanged: Rs. {wallet.balance}): {e}")
```

## ### Retail Inventory and Sales System

### Task:
Create a menu-driven inventory and sales application.

The program should allow the user to:
- Add stock
- View inventory
- Sell products
- Search for products

Handle invalid input, missing products, invalid quantities, and insufficient stock. Use `raise`, custom exceptions, `else`, and `finally` where appropriate.

The application should continue running after recoverable errors.

```python
class ProductNotFoundError(Exception):
  pass


class InvalidQuantityError(Exception):
  pass


class InsufficientStockError(Exception):
  pass


inventory = {
    "pen": {"price": 20, "stock": 50},
    "notebook": {"price": 150, "stock": 20},
}

while True:
  print("\n1. Add Stock | 2. View Inventory | 3. Sell Product | 4. Search | 5. Exit")
  choice = input("Choose option: ").strip()
  if choice == "5":
    break
  try:
    if choice == "1":
      name = input("Product name: ").strip().lower()
      qty = int(input("Quantity to add: "))
      if qty <= 0:
        raise InvalidQuantityError("Quantity must be positive.")
      if name not in inventory:
        price = float(input("New product price: "))
        inventory[name] = {"price": price, "stock": qty}
      else:
        inventory[name]["stock"] += qty
      msg = "Stock updated successfully."
    elif choice == "2":
      for k, v in inventory.items():
        print(f"{k}: Price={v['price']}, Stock={v['stock']}")
      msg = "Inventory displayed."
    elif choice == "3":
      name = input("Product to sell: ").strip().lower()
      if name not in inventory:
        raise ProductNotFoundError("Product not found.")
      qty = int(input("Quantity to sell: "))
      if qty <= 0:
        raise InvalidQuantityError("Quantity must be positive.")
      if qty > inventory[name]["stock"]:
        raise InsufficientStockError("Insufficient stock available.")
      inventory[name]["stock"] -= qty
      msg = f"Sold {qty} x {name}. Total: Rs. {inventory[name]['price'] * qty}"
    elif choice == "4":
      name = input("Search product name: ").strip().lower()
      if name not in inventory:
        raise ProductNotFoundError("Product not in inventory.")
      msg = f"Found {name}: {inventory[name]}"
    else:
      raise ValueError("Invalid menu choice.")
  except (
      ValueError,
      ProductNotFoundError,
      InvalidQuantityError,
      InsufficientStockError,
  ) as e:
    print(f"Error: {e}")
  else:
    print(msg)
  finally:
    print("Operation step completed.")
```

```python
class CustomerNotRegisteredError(Exception):
  pass


class RestaurantNotFoundError(Exception):
  pass


class ItemNotFoundError(Exception):
  pass


class InsufficientFoodStockError(Exception):
  pass


class MinimumOrderError(Exception):
  pass


class DeliveryAreaError(Exception):
  pass


class PaymentFailedError(Exception):
  pass


registered_customers = {"Ali": "North Karachi", "Sara": "Gulshan"}
restaurants = {
    "ABC Foods": {
        "areas": ["North Karachi", "Gulshan"],
        "min_order": 500,
        "menu": {"burger": {"price": 450, "stock": 10}},
    }
}


def place_food_order(customer_name, restaurant_name, item_name, qty, paid):
  if customer_name not in registered_customers:
    raise CustomerNotRegisteredError(
        f"Customer '{customer_name}' is not registered."
    )
  if restaurant_name not in restaurants:
    raise RestaurantNotFoundError(f"Restaurant '{restaurant_name}' not found.")
  rest = restaurants[restaurant_name]
  area = registered_customers[customer_name]
  if area not in rest["areas"]:
    raise DeliveryAreaError(
        f"Delivery not available in '{area}' for {restaurant_name}."
    )
  item_key = item_name.lower()
  if item_key not in rest["menu"]:
    raise ItemNotFoundError(f"Item '{item_name}' not available.")
  item = rest["menu"][item_key]
  if not isinstance(qty, int) or qty <= 0:
    raise ValueError("Quantity must be a positive integer.")
  if qty > item["stock"]:
    raise InsufficientFoodStockError(f"Only {item['stock']} left.")
  total = item["price"] * qty
  if total < rest["min_order"]:
    raise MinimumOrderError(f"Minimum order is Rs. {rest['min_order']}.")
  if not paid:
    raise PaymentFailedError("Payment was not completed.")
  item["stock"] -= qty
  return total


try:
  amount = place_food_order("Ali", "ABC Foods", "Burger", 2, True)
except Exception as e:
  print(f"Order failed: {e}")
else:
  print(
      f"Customer: Ali\nRestaurant: ABC Foods\nItem: Burger\nQuantity: 2\n->"
      f" Order placed successfully (Total: Rs. {amount})"
  )
finally:
  print("Food delivery order session closed.")
```
